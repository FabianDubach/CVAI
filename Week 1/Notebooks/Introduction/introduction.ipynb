```python
import numpy as np
import matplotlib.pyplot as pl
```

```python
img_rgb = pl.imread ('images/mh.jpg')
```

```python
pl.imshow (img_rgb)
```

```python
print (img_rgb.shape)
```

```python
img_gray = img_rgb [:, :, 0] * 1.0
```

```python
pl.imshow (img_gray, cmap = 'gray')
```

```python
print (img_gray)
```

```python
def f (x):
    y = x + 128
    y [y > 255] = 255
    return y
```

```python
x = np.linspace (0, 255, 256)
y = f (x)
pl.plot (x, y)
pl.xlim ([-10, 265])
pl.ylim ([-10, 265])
```

```python
neu = f (img_gray)
```

```python
print (neu)
```

```python
print (img_gray)
```

```python
img_gray [0, 0] = 0
img_gray [1, 0] = 255

neu [0, 0] = 0
neu [1, 0] = 255

pl.subplot (1, 2, 1); pl.imshow (img_gray, cmap = 'gray')
pl.subplot (1, 2, 2); pl.imshow (neu, cmap = 'gray')
```
